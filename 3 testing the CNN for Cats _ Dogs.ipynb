```python
from keras.models import Sequential
from keras.layers import Dense,Activation,Flatten
from keras.layers import Conv2D,MaxPooling2D,Dropout,BatchNormalization
from keras.optimizers import Adam

model=Sequential()

model.add(Conv2D(32,(3,3),input_shape=(50,50,1),kernel_initializer = 'orthogonal'))
model.add(BatchNormalization())
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size=(2,2)))
#The first CNN layer followed by Relu and MaxPooling layers

model.add(Conv2D(64,(3,3),kernel_initializer = 'orthogonal'))
model.add(BatchNormalization())
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size=(2,2)))
#The second convolution layer followed by Relu and MaxPooling layers

model.add(Conv2D(128,(3,3),kernel_initializer = 'orthogonal'))
model.add(BatchNormalization())
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size=(2,2)))



model.add(Flatten())
#Flatten layer to stack the output convolutions from third convolution layer
model.add(Dense(128,activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(64,activation='relu'))
model.add(Dropout(0.3))
model.add(Dense(2,activation='softmax'))
#The Final layer with two outputs for two categories

model.compile(loss='categorical_crossentropy',optimizer=Adam(learning_rate=0.0001),metrics=['accuracy'])        

```

```python
import os
import cv2
import numpy as np

test_path='test_imgs'

img_names=os.listdir(test_path) # test_imgs path eke thiyena hama path ektm ynava
#print(img_names)

category_dict={0:'Cat',1:'Dog'}

for img_name in img_names:          #image name vla one by one ynava
    
    img=cv2.imread(os.path.join(test_path,img_name))#image one by one read krl load karai
    h,w=img.shape[0:2]         # image eke height ekai weight ekai array ekkt ganima

    #preprocessing (As done for the training images)
    test_img=cv2.resize(img,(50,50))
    test_img=cv2.cvtColor(test_img,cv2.COLOR_BGR2GRAY)
    test_img=test_img/255
    test_img=test_img.reshape(1,50,50,1)
    
    results=model.predict(test_img)
    #print(results)
    
    label=np.argmax(results,axis=1)[0]
    acc=int(np.max(results,axis=1)[0]*100) #100 n vadi krl ena float number eke integer value eka gani
    
    #print(np.argmax(results,axis=1)[0])
    #print(results,label,category_dict[label],acc)
    
    category=category_dict[label]
    
    if(acc<75):
        
        category='NONE'

    #image  eke category ekai,accuracy ekai image eka matha place karai
    cv2.putText(img,category,(20,40),cv2.FONT_HERSHEY_SIMPLEX,1.5,(255,255,255),2)
    cv2.putText(img,str(acc)+'%',(20,80),cv2.FONT_HERSHEY_SIMPLEX,1.5,(255,255,255),2)
    
    cv2.imshow('LIVE',img)
    k=cv2.waitKey(0)
    if(k==27):                                 #escape eka ebuvhma images run vima sampurnayen navathiya haka
        break
cv2.destroyAllWindows()
```
